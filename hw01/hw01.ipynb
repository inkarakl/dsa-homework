6. Выручка (только оплаченные) по Алматы и по Астане отдельно.

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
almaty_total = 0
astana_total = 0

for c, a, p in zip(cities_clean, amounts_int, paid):
    if p:
        if c == "Almaty":
            almaty_total += a
        elif c == "Astana":
            astana_total += a
print(f"Almaty: {almaty_total}")  
print(f"Astana: {astana_total}")    
````

````python
max_sum = 0
max_city = ""
max_index = 0
i = 0

for c, a, p in zip(cities_clean, amounts_int, paid):
    if p and a > max_sum:
        max_sum = a
        max_city = c
        max_index = i
    i+= 1
print(max_index,max_city, max_sum)
````

````python
total = 0
weekend = 0 
for c, a, p,d in zip(cities_clean, amounts_int, paid,days):
    if p:
        total += 1
        if d in ("сб","вс"):
            weekend += 1

share = weekend / total * 100
print(round(share))
print( f"{share: .1f}%")
````

Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью while посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
kopilka = 15000
course = 400000
vsnos = 0
months = 0

while vsnos < course :
    vsnos += kopilka
    months +=1
    kopilka += 1000
print(months)
print(vsnos)

#без взноса он бы накопил за 27 месяцев 
````
